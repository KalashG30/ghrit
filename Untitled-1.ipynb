```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('deta.csv', on_bad_lines='skip')
data.shape
data.head()
```

```python
data.shape
```

```python
data['strength'].unique()
```

```python
data.isna().sum()
```

```python
data[data['password'].isnull()]
```

```python
data.dropna(inplace=True)
```

```python
sns.countplot(x='strength', data=data)
plt.show()
```

```python
password_tuple = np.array(data)
password_tuple
```

```python
import random
random.shuffle(password_tuple)
```

```python
x = [labels[0] for labels in password_tuple]
y = [labels[1] for labels in password_tuple]
```

```python
def word_split(inputs):
  character=[]
  for i in inputs:
    character.append(i)
  return character
```

```python
word_split('kzde5577')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer(tokenizer=word_split)
X = vectorizer.fit_transform(x)
```

```python
X.shape
```

```python
vectorizer.get_feature_names_out()
```

```python
first_vector_document = X[0]
first_vector_document
```

```python
first_vector_document.T.todense()
```

```python
tf_idf = pd.DataFrame(first_vector_document.T.todense(), index=vectorizer.get_feature_names_out(), columns=['TF-IDF'])
tf_idf.sort_values(by=['TF-IDF'], ascending=False)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
X_train.shape
```

```python
type(X)
```

```python
print(y)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn import metrics
```

```python
models = {
    'Random Forest': RandomForestClassifier(criterion = 'entropy', random_state = 0)
}
```

```python
for model_name, clf in models.items():
  clf.fit(X_train, y_train)
  y_pred = clf.predict(X_test)
```

```python
prediction_on_training_data = clf.predict(X_train)
accuracy = metrics.accuracy_score(y_train, prediction_on_training_data)
```

```python
print('Accuracy on training data : ', accuracy)
```

```python
prediction_on_test_data = clf.predict(X_test)
accuracy = metrics.accuracy_score(y_test, prediction_on_test_data)
```

```python
print('Accuracy on test data : ', accuracy)
```

```python
import joblib
joblib.dump(models['Random Forest'], "password_strength_model_1.pkl")
joblib.dump(vectorizer, "tfidf_vectorizer_1.pkl")
```
